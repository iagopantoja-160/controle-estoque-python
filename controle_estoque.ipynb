```python
# ==========================================
# SISTEMA DE CONTROLE DE ESTOQUE EM PYTHON
# Versão 7 - Salvamento em CSV
# ==========================================

import csv
from pathlib import Path

estoque = []

LIMITE_ESTOQUE_BAIXO = 5

ARQUIVO_CSV = Path(
    "/content/drive/MyDrive/controle-estoque-python/estoque.csv"
)


# SALVAR ESTOQUE EM CSV
def salvar_estoque():
    with open(ARQUIVO_CSV, "w", newline="", encoding="utf-8") as arquivo:
        campos = ["nome", "quantidade", "preco"]

        escritor = csv.DictWriter(arquivo, fieldnames=campos)
        escritor.writeheader()
        escritor.writerows(estoque)


# CARREGAR ESTOQUE DO CSV
def carregar_estoque():
    if not ARQUIVO_CSV.exists():
        print("Nenhum arquivo anterior encontrado.")
        return

    try:
        with open(ARQUIVO_CSV, "r", newline="", encoding="utf-8") as arquivo:
            leitor = csv.DictReader(arquivo)

            produtos_carregados = []

            for linha in leitor:
                produto = {
                    "nome": linha["nome"],
                    "quantidade": int(linha["quantidade"]),
                    "preco": float(linha["preco"])
                }

                produtos_carregados.append(produto)

        estoque.clear()
        estoque.extend(produtos_carregados)

        print("Estoque carregado com sucesso!")

    except (OSError, ValueError, KeyError, csv.Error) as erro:
        print(f"Não foi possível carregar o estoque: {erro}")


# 1 - CADASTRAR PRODUTO
def cadastrar_produto():
    print("\n--- CADASTRAR PRODUTO ---")

    nome = input("Nome do produto: ").strip()

    if not nome:
        print("O nome não pode ficar vazio.")
        return

    for produto in estoque:
        if produto["nome"].lower() == nome.lower():
            print("Esse produto já está cadastrado.")
            return

    try:
        quantidade = int(input("Quantidade: "))
        preco = float(input("Preço unitário: R$ "))

        if quantidade < 0 or preco < 0:
            print("Quantidade e preço não podem ser negativos.")
            return

    except ValueError:
        print("Digite valores numéricos válidos.")
        return

    produto = {
        "nome": nome,
        "quantidade": quantidade,
        "preco": preco
    }

    estoque.append(produto)
    salvar_estoque()

    print("Produto cadastrado e salvo com sucesso!")


# 2 - CONSULTAR ESTOQUE
def consultar_estoque():
    print("\n--- ESTOQUE ATUAL ---")

    if not estoque:
        print("Nenhum produto cadastrado.")
        return

    for produto in estoque:
        print(
            f"Produto: {produto['nome']} | "
            f"Quantidade: {produto['quantidade']} | "
            f"Preço: R$ {produto['preco']:.2f}"
        )


# 3 - ATUALIZAR QUANTIDADE
def atualizar_quantidade():
    print("\n--- ATUALIZAR QUANTIDADE ---")

    nome = input("Nome do produto: ").strip()

    for produto in estoque:
        if produto["nome"].lower() == nome.lower():

            try:
                nova_quantidade = int(
                    input("Digite a nova quantidade: ")
                )

                if nova_quantidade < 0:
                    print("A quantidade não pode ser negativa.")
                    return

                produto["quantidade"] = nova_quantidade
                salvar_estoque()

                print("Quantidade atualizada e salva!")
                return

            except ValueError:
                print("Digite um número inteiro válido.")
                return

    print("Produto não encontrado.")


# 4 - EXCLUIR PRODUTO
def excluir_produto():
    print("\n--- EXCLUIR PRODUTO ---")

    nome = input("Nome do produto: ").strip()

    for produto in estoque:
        if produto["nome"].lower() == nome.lower():

            confirmacao = input(
                f"Excluir {produto['nome']}? (s/n): "
            ).strip().lower()

            if confirmacao == "s":
                estoque.remove(produto)
                salvar_estoque()
                print("Produto excluído e alteração salva!")
            else:
                print("Exclusão cancelada.")

            return

    print("Produto não encontrado.")


# 5 - ALERTA DE ESTOQUE BAIXO
def verificar_estoque_baixo():
    print("\n--- ALERTA DE ESTOQUE BAIXO ---")

    produtos_baixos = [
        produto for produto in estoque
        if produto["quantidade"] <= LIMITE_ESTOQUE_BAIXO
    ]

    if not produtos_baixos:
        print("Nenhum produto com estoque baixo.")
        return

    for produto in produtos_baixos:
        if produto["quantidade"] == 0:
            situacao = "SEM ESTOQUE"
        else:
            situacao = "ESTOQUE BAIXO"

        print(
            f"ALERTA: {produto['nome']} | "
            f"Quantidade: {produto['quantidade']} | "
            f"Situação: {situacao}"
        )


# 6 - CALCULAR VALOR TOTAL
def calcular_valor_total():
    print("\n--- VALOR TOTAL DO ESTOQUE ---")

    if not estoque:
        print("Nenhum produto cadastrado.")
        return

    valor_total = 0

    for produto in estoque:
        subtotal = produto["quantidade"] * produto["preco"]
        valor_total += subtotal

        print(
            f"Produto: {produto['nome']} | "
            f"Subtotal: R$ {subtotal:.2f}"
        )

    print("------------------------------")
    print(f"VALOR TOTAL: R$ {valor_total:.2f}")


# MENU PRINCIPAL
def menu():
    carregar_estoque()

    while True:
        print("\n==============================")
        print(" SISTEMA DE CONTROLE DE ESTOQUE")
        print("==============================")
        print("1 - Cadastrar produto")
        print("2 - Consultar estoque")
        print("3 - Atualizar quantidade")
        print("4 - Excluir produto")
        print("5 - Verificar estoque baixo")
        print("6 - Calcular valor total")
        print("0 - Sair")

        opcao = input("Escolha uma opção: ")

        if opcao == "1":
            cadastrar_produto()

        elif opcao == "2":
            consultar_estoque()

        elif opcao == "3":
            atualizar_quantidade()

        elif opcao == "4":
            excluir_produto()

        elif opcao == "5":
            verificar_estoque_baixo()

        elif opcao == "6":
            calcular_valor_total()

        elif opcao == "0":
            print("Sistema encerrado.")
            break

        else:
            print("Opção inválida. Tente novamente.")


# INICIAR SISTEMA
menu()
```
